```python
import BM25
import pandas as pd

df = pd.read_csv("sections.csv")
# df
# row_val = df.iloc[1]['Header']
# row_val
# print(row_val)

# load csv file
corpus = BM25.load("sections.csv", document_column="Text")

# build the search index
retriever = BM25.index(corpus)

# get top 5 results
queries = ["Welk gedrag heeft iemand met 1p36?"]
results = retriever.search(queries, k=5)

# print the top results for the first query
for result in results[0]:
    print(f"Score: {result['score']:.2f} | Document: {result['document']} | Header: {df.iloc[result['id']]['Header']}")
    print("\n")
```
